```python
import sys
import importlib.util

print("Python:", sys.version)
print("Executable:", sys.executable)

packages = [
    "numpy",
    "pandas",
    "sklearn",
    "xgboost",
    "torch",
    "gymnasium",
    "stable_baselines3",
    "fastapi",
    "mlflow",
]

for package in packages:
    print(f"{package}:",
          "OK" if importlib.util.find_spec(package) else "MISSING")
```

```python
import sys
import importlib

print("Python:", sys.version)
print("Executable:", sys.executable)

packages = [
    "numpy",
    "pandas",
    "sklearn",
    "xgboost",
    "torch",
    "gymnasium",
    "stable_baselines3",
    "fastapi",
    "mlflow",
]

for package in packages:
    try:
        importlib.import_module(package)
        print(f"{package}: OK")
    except Exception as e:
        print(f"{package}: ERROR - {e}")
        
```
